```python
#2
def get_bill_total(bill_am,sale_tax):
  amount=bill_am*(1+sale_tax/100)
  return float("{:.2f}".format(amount))
```

```python
bill=float(input("Enter bill amount: "))
tax = float(input("Enter sale tax between 0 and 100: "))
while not(0<=tax<=100):
  print(tax,"must be between 0 and 100")
  tax = float(input("Enter sale tax: "))
print(get_bill_total(bill,tax))
```

```python
#3
def print_perfect_squares(k):
  for p in range(1,k+1):
    if p*p <=k:
      print(p*p, end=" ")
```

```python
num = int(input("Enter an integer: "))
print_perfect_squares(num)
```

```python
#4
def add_up_range(num1,num2,num3):
  return sum(list(range(num1,num2,num3)))
```

```python
start_loc=int(input("Enter starting location: "))
end_loc = int(input("Enter end location: "))
step = int(input("Enter step: "))
print(add_up_range(start_loc,end_loc,step))
```

```python
#4 OR
def add_up_range_1(num1,num2,num3):
  tot =0
  for num in range(num1,num2,num3):
    tot+=num
  return tot
```

```python
start_loc=int(input("Enter starting location: "))
end_loc = int(input("Enter end location: "))
step = int(input("Enter step: "))
print(add_up_range_1(start_loc,end_loc,step))
```

```python
#5

def phi(k):
  from math import gcd
  count=0
  if k==0:
    return "Please enter a not null integer!"
  if k==1:
    return 0
  for p in range(1,k+1):
    if gcd(p,k)==1:
      count+=1
  return count
```

```python
num = int(input("Enter num: "))
print(phi(num))
```

```python
#6
def is_upper(stg):
  if stg==stg.upper():
    return "Every letter of the string is uppercase."
  else:
    return "There is a letter that is not uppercase."

string=input("Enter a string of letters: ")
while not(string.isalpha()):
  print("String must be all letters")
  string=input("Enter a string of letters: ")

is_upper(string)
```

```python
#7 If we want possible repeated characters
def rand_lower_string(k):
  from string import ascii_lowercase
  from random import choice
  s=ascii_lowercase
  return "".join([choice(s) for l in range(k)])
```

```python
num=int(input("Enter an integer: "))
print(rand_lower_string(num))
```

```python
#7 If we want distinct characters
def rand_lower_string1(k):
  from string import ascii_lowercase
  from random import sample
  s=ascii_lowercase
  return "".join(sample(list(s),k))
```

```python
num=int(input("Enter an integer: "))
print(rand_lower_string1(num))
```

```python
#8
def simplify_text(stg):
  from string import punctuation
  stg=stg.lower()
  for c in "'-":
    punctuation=punctuation.replace(c,"")
  for char in punctuation:
    stg=stg.replace(char,"")
  return stg
```

```python
string = input("Enter a string: ")
print(simplify_text(string))
```

```python
#9
def  average(L):
  return float("{:.2f}".format(sum(L)/len(L)))
```

```python
num=int(input("Enter number of items in list: "))
list_num = [float(input("Enter numbers: ")) for k in range(num)]
print(average(list_num))
```

```python
#10
def string_sort(stg):
  L=list(stg)
  L.sort()
  return "".join(L)
```

```python
string=input("Enter string: ")
string=string.lower()
while not(string.isalpha()):
  print("String must be all letters.")
  string=input("Enter string: ")

print(string_sort(string))
```

```python
#11
def integer_lengths(L):
  return sum([len(str(item)) for item in L])
```

```python
num=int(input("Enter number of items in list: "))
list_integers=[int(input("Enter an integer: ")) for k in range(num)]
print(integer_lengths(list_integers))
```

```python
#12
def union(L1,L2):
  Union=[]
  L1.extend(L2)
  for item in L1:
    if item not in Union:
      Union.append(item)
  return Union
```

```python
num1 = int(input("Enter number of elements in the first list: "))
List1 = [input("Enter elements of list 1: ") for k in range(num1)]

num2 = int(input("Enter number of elements in the second list: "))
List2 = [input("Enter elements of list 2: ") for k in range(num2)]

print(List1)
print(List2)
print(union(List1,List2))
```

```python
#14
def changes_by_one(L):
  result=[]
  for k in range(len(L)-1):
    if L[k+1]-L[k]==1:
      result.append(k+1)
  return result
```

```python
num=int(input("Number of items in list: "))
num_list=[int(input("Enter a number: ")) for k in range(num)]
print(changes_by_one(num_list))
```

```python
#14
def changes_by_one1(L):
  return [k+1 for k in range(len(L)-1) if L[k+1]-L[k]==1]
```

```python
num=int(input("Number of items in list: "))
num_list=[int(input("Enter a number: ")) for k in range(num)]
print(changes_by_one1(num_list))
```

```python
def changes_by_one2(L):
  return [k for k in range(1,len(L)) if L[k]-L[k-1]==1]

n = int(input("Enter number of elements in the first list: "))
num_List = [int(input("Enter numbers: ")) for k in range(n)]

print(changes_by_one2(num_List))
```
