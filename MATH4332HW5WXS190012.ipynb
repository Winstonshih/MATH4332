```python
#5
lowercaseletters = input("Enter a string of lowercase letters: ")
wordfreq = {}
for letter in lowercaseletters:
  wordfreq[letter] = 0
  if letter in wordfreq:
    wordfreq[letter] += 1
with open('/content/drive/MyDrive/MATH4332HW5files/wordlist.txt') as f:
    words = [line.strip() for line in f]
    for line in words:
        word_count = {}
        for letter in line:
            if letter in word_count:
                word_count[letter] += 1
            else:
                word_count[letter] = 1
        properword = True
        for letter, count in word_count.items():
            if letter not in wordfreq or wordfreq[letter] < count:
                properword = False
                break
        if properword:
            print(line)
```
