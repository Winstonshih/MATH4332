```python
#Winston Shih (WXS190012)
#1
import random
from random import randint
questions=["What is the capital of Texas? ",
"What is the largest city in Texas? ", "Who is the Governor of Texas? ",
"What is the most populated county in Texas? ",
"Which city is UT Dallas located in? "]
answers=["Austin", "Houston", "Greg Abbott", "Harris", "Richardson"]
count=0
for i in range(3):
  randomquestion=randint(0,4)
  question=input(questions[randomquestion])
  if(question==answers[randomquestion]):
    print("You got the question right!")
    count+=1
  else:
    print("You got the question wrong!")
print("You got",count," out of 3 questions right!")
```

```python
#4b
import random
from random import shuffle
sentence=input("Enter a sentence: ")
sentence=sentence.lower()
L=sentence[:-1].split()
random.shuffle(L)
L[0]=L[0][0].upper()+L[0][1:]
s1=L[0]+" "+" ".join(L[1:])+"."
print("Scrambled sentence:",s1)
```
