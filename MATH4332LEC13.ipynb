```python
# Examples
# Find all palindromic numbers between 10 and 10000
# that is the numbers that is the same backward as forward
# e.g.: 121, 2332
L = []
for k in range(10,10001):
  s = str(k)    #  str function is used here to turn the integer k into
                # a string so we can use slices to reverse it.
  if s == s[::-1]:
    L.append(k)
print(L)
```

```python
# Short circuiting

# What is short-circuiting in Python?

# Short-circuiting in Python is a technique by which execution
# of boolean expression containing 'or' and 'and' operator in Python
# is halted as soon as the truth value of the expression is determined.


# How does short-circuiting occur for 'and' operator?
# As soon as the interpreter finds a statement that is false,
# it will short circuit and no longer check the other statements
# in the expression.

# How does short-circuiting occur for 'or' operator?

# As soon as the interpreter finds a statement that is true,
# it will short circuit and no longer check the other statements
# in the expression.

# What is the use of short-circuiting in Python?

# It makes the computation faster by not checking the other statements
# once the truth value is established.
```

```python
# w is a list of words. We are are searching for words
# in the list whose fourth character is "x"
# we might try the following

s = input("Enter a string: ")
from string import punctuation
for c in punctuation:
  s=s.replace(c,"")
w = s.split()
print(w)
for word in w:
  if word[3] == "x":# this can lead to string index out of range error
    print(word)
```

```python
# How to fix the above program?
for word in w:
  if len(word)>=4 and word[3] == "x": #len(word) will be executed first
    print(word)           # if it fails, word[3] will not be checked
```

```python
# The key to why this works is short circuiting.
# Python starts by checking the first part of the condition,
# len(word)>=4. If that condition turns out to be false,
# then the whole and condition is guaranteed to be false,
# and so there is no point in even looking at the second condition.
# So Python doesn’t bother with the second condition.
```

```python
# Short-circuiting also happens with or conditions.
# In this case, Python checks the first part of the or
# and if it is true, then the whole or is guaranteed to be true
# and so Python will not bother checking the second part of the or.
```

```python
# Observe that the problem remains when word[3]== "x" is written first
# in the conditional statements
for word in w:
  if word[3] == "x" and len(word)>=4 :
    print(word)
```

```python
#Formatting for integers
#To format integers, the formatting code is {:d}.
#Putting a number in front of the d allows us to right justify integers.
print("{:4d}".format(3))
print("{:4d}".format(13))
print("{:4d}".format(123))
print("{:4d}".format(1234))
# The number 4 in these examples says that
# the value is allotted 4 spots.
```

```python
# {:<d} is used to left-justify integers
print("{:<4d}".format(3))
print("{:<4d}".format(13))
print("{:<4d}".format(123))
print("{:<4d}".format(1234))
```

```python
# {:^d} is used to center integers
print("{:^5d}".format(3))
print("{:^5d}".format(123))
print("{:^5d}".format(12345))
# Each of these allots five spaces for the integer
# and centers it within those five spaces.
```

```python
# {:,d} is used to format the integers with commas
print("{:,d}".format(10000000000))
```

```python
# Formatting floats
# {:f} is used to format float
# {:.2f} is used to only display the number to two decimal places
# The 2 can be changed to change the number of decimal places.
# E.g
num = 3.14356797454
print("{:.3f}".format(num))
print("{:.5f}".format(num))
```

```python
# E.g
# Pythagorean triples: triple of whole numbers (x,y,z)
# such that x^2 + y^2 = z^2
# Here is a program that finds all the Pythagorean triples (x,y,z)
# where x, y, and z are positive and less than 25.
# x,y,z are positive integers and 1<=x,y,z<=24
L = []
for x in range(1,25):
  for y in range(1,25):
    for z in range(1,25):
      if x**2 + y**2 == z**2:
        L.append([x,y,z])
print(L)
```

```python
# How to get rid off redundant solutions?
# E.g of redundant solutions: (3,4,5) and (4,3,5), (6,8,10) and (8,6,10)
L = []
for x in range(1,25):
  for y in range(x,25):
    for z in range(y,25):
      if x**2 + y**2 == z**2:
        L.append([x,y,z])
print(L)
```

```python
# Another Approach
L = []
for x in range(1,25):
  for y in range(1,25):
    for z in range(1,25):
      if x**2 + y**2 == z**2:
        L.append([x,y,z])
#print(L)
L2 = []
for item in L:
  item.sort()
  if item not in L2:
    L2.append(item)
print(L2)
```

```python
# How to remove triples which are multiple of each other like
# (3,4,5), (6,8,10), (9,12,15) and (12,16,20)?
# The following program finds only primitive Pythagorean triples,
# ie those that aren’t multiples of another triple.
L = []
for x in range(1,25):
  for y in range(x,25):
    for z in range(y,25):
      if x**2 + y**2 == z**2:
        for k in range(2,x):
          if x%k==0 and y%k==0 and z%k==0:
            break
        else:
          L.append([x,y,z])

print(L)
```
