```python
#Winston Shih
#1a.
import numpy as np
pos_term_denominators=np.arange(1,3331,4)
pos_term=1/pos_term_denominators
neg_term=1/(pos_term_denominators+2)
print(4*(np.sum(pos_term)-np.sum(neg_term)))
#The approximation is accurate by 2 decimal points.
```

```python
#1b.
import numpy as np
even=np.arange(2,55555,2)
one_below_even=even-1
one_above_even=even+1
print(2*np.prod(even/one_below_even)*np.prod(even/one_above_even))
#The approximation is accurate by 4 decimal points
```

```python
#2.
import numpy as np
import matplotlib.pyplot as plt
def my_fun_plotter(f,x):
  plt.plot(x,f(x),'k')
  plt.xlabel('x')
  plt.ylabel('f(x)')
  plt.title("My_fun_plotter")
  plt.show()
my_fun_plotter(lambda x: np.sqrt(x) + np.exp(np.sin(x)), np.linspace(0,
                                                              2*np.pi,100))
```

```python
#3
import numpy as np
import matplotlib.pyplot as plt
def my_poly_plotter(n,x):
  for i in range(1,n+1):
    colors = ['k', 'g', 'r', 'c', 'm', 'y']
    plt.plot(x,x**i,colors[i-1], label=f'x^{i}')
my_poly_plotter(5, np.linspace(-1, 1, 200))
plt.legend()
plt.xlabel('x')
plt.ylabel('pk(x)=x^k')
plt.title("My_poly_plotter")
plt.show()
```

```python
#4
import numpy as np
import matplotlib.pyplot as plt
def my_surface_plotter(f,x,y, option):
  X, Y = np.meshgrid(x, y)
  Z=f(X,Y)
  if(option=='surface'):
    fig = plt.figure(figsize=(12, 6))
    ax = fig.add_subplot(1, 2, 1, projection='3d')
    ax.plot_surface(X,Y,Z)
    ax.set_title('Surface plot')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x,y)=cosy*sinx', labelpad=-0.7)
  elif(option=='contour'):
    fig = plt.figure(figsize=(12,6))
    ax = fig.add_subplot(1, 2, 2)
    ax.contour(X,Y,Z,cmap='viridis')
    ax.set_title('Contour plot')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.contour(X,Y,Z)
f = lambda x, y: np.cos(y)*np.sin(np.exp(x))
my_surface_plotter(f, np.linspace(-1, 1, 20), np.linspace(-2, 2, 40),
                   'surface')
my_surface_plotter(f, np.linspace(-1, 1, 20), np.linspace(-2, 2, 40),
                   'contour')
```

```python
#5
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-4,4,100)
y=np.linspace(-3,3,100)
X,Y=np.meshgrid(x,y)
Z=X*Y*(X**2-Y**2)/(X**2+Y**2)
fig = plt.figure(figsize=(12,6))
ax = fig.add_subplot(1, 2, 2, projection='3d')
ax.plot_wireframe(X,Y,Z)
ax.set_title('Wireframe plot')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z',labelpad=-4)
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(X,Y,Z)
ax.set_title('Surface plot')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z', labelpad=-0.7)
plt.tight_layout()
plt.show()
```
