```python
#4
coursebook={}
for i in range(5):
  person=input("Enter a name: ")
  courses=input("Enter the courses you took (use commas): ")
  coursebook[person]=courses.split(",")
search=input("Enter the course you want to search: ")
print("The people who took "+search+" are:")
for person in coursebook:
  if search in coursebook[person]:
    print(person)
```

```python
#6
from random import shuffle
dict={"Who won the Super Bowl 2025?": "Philadelphia Eagles",
      "Who played Sergeant Doakes?": "Erik King",
      "Who created Python?": "Guido van Rossum"}
d=list(dict)
shuffle(d)
for key in d:
  print(key)
  answer=input("Enter your answer: ")
  if(answer==dict[key]):
    print("Correct!")
  else:
    print("Incorrect!")
    print("The right answer is:",dict[key])
```
