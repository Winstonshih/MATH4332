```python
# while loop

# For loop allows us to repeat things a specified number of times.
# while loop is used when we need to repeat something,
# but we don’t know ahead of time exactly how many times
# it has to be repeated.
```

```python
# A while loop or indefinite loop is a set of instructions
# that is repeated as long as the associated logical expression is true.
# The following is the abstract syntax of a while loop block.
```

```python
# CONSTRUCTION: While Loop
# while <logical expression>:
    # Code block to be repeated until logical statement is false
    # code block
```

```python
# When Python reaches a while loop block, it first determines if the logical
# expression of the while loop is true or false. If the expression is true,
# the code block will be executed, and after it is executed,the program returns
# to the logical expression at the beginning of the while statement.
# If it is false, then the while loop will terminate.
```

```python
# Determine the number of times 8 can be divided by 2 until the
# result is less than 1.
count = 0
n = 8
while n >= 1:
  n /= 2  # n = n/2
  count += 1
print(count)
```

```python
# Temperature converter program:
# For a temperature converter program, earlier in this class,
# we used the following:
temp = float(input('Enter a temperature in Celsius: '))
print('In Fahrenheit, that is', 9/5*temp+32)
# In the above program, the user has to restart the program
# for every new temperature.
```

```python
# A while loop will allow the user to repeatedly enter temperatures.
# like below:
temp = 100 # (Any value !=-500)
while temp !=-500:
  temp = float(input("Enter temp in C(-500 to quit): "))
  print("Temp in F is", temp*9/5 + 32)
```

```python
# A while loop is a lot like an if statement.
# The difference is that the indented statements in an if block
# will only be executed once, whereas the indented statements
# in a while loop are repeatedly executed.
```

```python
# One problem with the previous program is that when the user enters
# in −500 to quit, the program still converts the value −500
# and doesn’t give any message to indicate that the program has ended.
# A nicer way to do the program is shown below.

temp = 100
while temp !=-500:
  temp = float(input("Enter temp in C(-500 to quit): "))
  if temp!=-500:
    print("Temp in F is", temp*9/5 + 32)
  else:
    print("Bye")
```

```python
# Recall:
# Guessing game program with only one guess.
from random import randint
secrete_num = randint(1,5)
print(secrete_num)
guess = int(input("Enter an integer between 1 and 5: "))
if guess == secrete_num:
  print("You won.")
else:
  print("You lost.")
```

```python
# A program that gets input from a user may want to check
# to see that the user has entered valid data.
# The smallest possible temperature is absolute zero, -273.15 degree C.
# The program below takes absolute zero into account:

temp = float(input("Enter a temp in C: "))

if temp < -273.15:
  print("Enter a valid temp. Temp must be at least -273.15")
else:
  print("Temp in F is", temp*9/5 + 32)
```

```python
# One way to improve this is to allow the user to keep reentering
# the temperature until they enter a valid one.

temp = float(input("Enter a temp in C: "))
while temp < -273.15:
  print("Enter a valid temp. Temp must be at least -273.15")
  temp = float(input("Enter a temp in C: "))

print("Temp in F is", temp*9/5 + 32)
```

```python
# Which while-loop causes an infinite loop?
# Example 1
n = 1
count = 0
while n > 0:
  n /= 2 # n=n/2
  count +=1
print(count)
# The above will not infinite loop because eventually n will be so small that
# Python cannot tell the difference between n and 0.
```

```python
# Example 2
n = 2
while n > 0:
  if n % 2 == 0:
      n += 1
  else:
      n -= 1
# The above example will infinite loop because n will oscillate
# between 2 and 3 indefinitely.
```

```python
# The break statement
# It can be used to break out of a for or while loop
# before the loop is finished.

# Here is a program that allows the user to enter up to 3 numbers.
# The user can stop early by entering a negative number.

for k in range(3):
  num = float(input("Enter an integer: "))
  if num < 0:
    break
```

```python
# The above could also be accomplished with a while loop.
k = 0
while k < 3:
  num = float(input("Enter a number: "))
  if num < 0:
    break
  k+=1
```

```python
# The else statement

# In a for loop, there is an optional else that you can use
# with break statement.
# The indented code under else gets executed only
# if the loop completes without a break happening.
for k in range(3):
  num = float(input("Enter an integer: "))
  if num < 0:
    print("This an early stop.")
    break
else:
  print("The user enters all values")
```

```python
# Here are two ways to check if an integer num is prime.

# The for loop approach
num = int(input("Enter a number: "))
for k in range(2,num):
  if num%k ==0:
    print(num,"is not prime")
    break
else:
  if num >2:
    print(num,"is prime")
  else:
    print(num,"is not prime")
```

```python
# The while loop approach
num = int(input("Enter a number: "))
k = 2
while k < num and num%k!=0:
  k+=1

print(k)
if k == num:
  print(num,"is prime")
else:
  print(num,"is not prime")
```

```python
#4
num1=int(input("Enter number 1: "))
num2=int(input("Enter number 2: "))

while (num1 <=0) or (num2 <=0):
  num1 = int(input("Enter number 1: "))
  num2 = int(input("Enter number 2: "))

if num1>num2:
  while num2!=0:
    r=num1%num2
    num1,num2=num2,r
  print("The gcd is",num1)
elif num1<num2:
  while num1!=0:
    r=num2%num1
    num2,num1=num1,r
  print("The gcd is",num2)
else:
  print("The gcd is",num2)
```

```python
#5
from random import randint
from pprint import pprint
L=[[0]*5 for k in range(5)]
pprint(L)
j=0
while j<10:
  row = randint(0,4)
  col = randint(0,4)
  if L[row][col]==0:
    L[row][col]=1
    j+=1
pprint(L)
```

```python
#5
from random import randint
L = [[0]*5 for k in range(5)]
from pprint import pprint
pprint(L)
count = 0
while count < 10:
  row = randint(0,4)
  col = randint(0,4)
  if L[row][col]==0:
    L[row][col]=1
  else:
    count=count-1

  count+=1
pprint(L)
```
