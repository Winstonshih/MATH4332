```python
# The code works, but it is very tedious.
# Tedious code like this can often be greatly simplified
# with lists and loops:
Questions =["What is the capital of Togo?: ", "What state are we in?: "]
Answers = ["Lome","Texas"]
num_correct_ans=0
for k in range(len(Questions)):
  guess=input(Questions[k])
  if guess.lower()==Answers[k].lower():
    print("You got it right.")
    num_correct_ans+=1
  else:
    print("You got it wrong.")
  print("You have",num_correct_ans,"out of",k+1,"right.")
```

```python
# More with Lists
# List and random module
# There are some nice functions in the random module that work on lists.
# choice(L): picks a ramdom item from L
# sample(L,n): picks group of n random item from L
# shuffle(L): shuffles item from L
# The shuffle function modifies the original list
# A copy of the list L is needed before using shuffle
# if you don’t want your original list changed.
```

```python
# E1: Use choice to pick an element from a list
from random import choice
names = ["Sam", "Paul","Anna","Albert"]
rand_item = choice(names)
print(rand_item)
```

```python
#E2: Use sample to pick several.
from random import sample
names = ["Sam", "Paul","Anna","Albert"]
rand_sev_items = sample(names,3)
print(rand_sev_items)
```

```python
# E3: We can use choice to pick a character from a string as well
from string import ascii_lowercase
s=ascii_lowercase
for k in range(5):
  r = choice(s)
  print(r,end="")
```

```python
# Or
st = ""
for k in range(5):
  st = st + choice(s)
print(st)
```

```python
# E4: Shuffle is used to pick a random ordering
from random import shuffle
names = ["Sam", "Paul","Anna","Albert"]
shuffle(names)
print(names)
for name in names:
  print(name)
```

```python
# What do we do if we want questions to come in
# random orders in the quiz game?
Questions =["What is the capital of Togo?: ", "What state are we in?: ","2+3: "]
Answers = ["Lome","Texas","5"]
num_correct_ans=0
# List of indices of the questions
L=[]
for k in range(len(Questions)):
  L.append(k)
shuffle(L)
for k in range(len(Questions)):
  guess=input(Questions[L[k]])
  if guess.lower()==Answers[L[k]].lower():
    print("You got it right.")
    num_correct_ans+=1
  else:
    print("You got it wrong.")
  print("You have",num_correct_ans,"out of",k+1,"right.")
```

```python
# split method: It returns a list of the words of a string
# The method assumes that words are separated by whitespace
# which can be spaces, tab or newline characters
```

```python
# E.g
s = "Hello! Is the homework due tonight?"
L = s.split()
print(L)
```

```python
# Remove punctuations from s before splitting it
from string import punctuation
s = "Hello! Is the homework due tonight?"
for c in punctuation:
  s = s.replace(c,"")
L = s.split()
print(L)
```

```python
# split breaks up the string at spaces
s = "t\ny\th u"
L = s.split()
print(L)
```

```python
# Program that counts how many times a certain word occurs in a string.
s= "Hello! Is the homework due tonight? hello"
for c in punctuation:
  s=s.replace(c,"")
s=s.lower()
L = s.split()
word = input("Enter word: ")
word=word.lower()
print(word, 'appears', L.count(word), 'times.')
```

```python
# Optional argument for split:
# The split method takes an optional argument that allows it to break
# the string at places other than spaces. Here is an example:
s = "1-222-777-6754"
L = s.split("-")
print(L)
```

```python
# Program that creates anagram from a given word
# An anagram of a word uses the same letters as
# the word but in a different order.
s = "idle"
# we can use list(s) to create a list of characters of s
letter_list = list(s)
shuffle(letter_list)
anagrams = "".join(letter_list)
print(anagrams)
```

```python
# What if item[1] does not exist?
m = ["One", "Two", "Three","Four","Five","u"]
L6 = [item[1] for item in m if len(item) >=2]
print(L6)
```

```python
m = ["One", "Two", "Three","Four","Five","u"]
L6 = [item[1] for item in m if len(item)==4]
print(L6)
```

```python
# The above corresponds to the following long way
L=[]
for item in m:
  if len(item)==4:
    L.append(item[1])
print(L)
```

```python
# Multiple for loops in list comprehensions
L = [[k,l] for k in range(3) for l in range(3)]
print(L)
```

```python
# Usual way
K = []
for k in range(3):
  for l in range(3):
    K.append([k,l])
print(K)
```

```python
# Using list comprehensions
# Use list comprehension to generate 10 random integers
# between 1 and 25
from random import randint
L = [randint(1,25) for k in range(10)]
print(L)
```

```python
E = [1,2,3,4,5]
# Use list comprehension to create a list that replaces
# every element of E by its cube
L = [item**3 for item in E]
print(L)
```

```python
# Use list comprehension to count how many items
# in E are greater than 3
# Solution 1
len([item for item in E if item > 3])
```

```python
# Solution 2
sum([1 for item in E if item > 3])
```

```python
# Let L be a list of numbers between 1 and 100
L=[randint(1,100) for k in range(150)]
print(L)
# create a new list whose first element is how many ones are in L,
# whose second element is how many twos are in L, etc.
new_list = [L.count(k) for k in range(1,101)]
print(new_list)
```

```python
# The join method can often be used with list comprehensions
# to quickly build up a string.
from string import ascii_lowercase
from random import choice
s = ascii_lowercase
t = "".join([choice(s) for k in range(10)])
print(t)
```

```python
# Flipping orders of entries in the list
L = [[3,7],[9,2],[7,22]]
flip = [[b,a] for a,b in L]
print(flip)
```
