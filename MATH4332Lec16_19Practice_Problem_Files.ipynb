```python
#4
with open("/content/drive/MyDrive/Ex_text_files/Ex_text_files/wordlist.txt") as f:
  lines = [line.rstrip() for line in f]
```

```python
#a
L=[]
for word in lines:
  if "quot" in word:
    L.append(word)
print(len(L))
print(L)
```

```python
#b
L=[]
for word in lines:
  for c in "aeiou":
    if c not in word:
      break
  else:
    L.append(word)

print(100*len(L)/len(lines))
```

```python
#b
L=[]
for word in lines:
  for vowel in list("aeiou"):
    if vowel not in word:
      break
  else:
    L.append(word)

print("Percentage: {:.2f}".format(100*len(L)/len(lines)))
```

```python
#c
print("Average word length: {:.2f}".format(sum([len(word) for word in lines])/len(lines)))
```

```python
#d
L=[]
for word in lines:
  if len(word)>=6 and word[:3]==word[-3:]:
    L.append(word)
print(L)
```

```python
#e
longest = 0
for word in lines:
  if word[0]==word[-1] and len(word)>longest:
    longest=len(word)
    long_word=word
print("The longest word is: {}.\nIts length is: {}".format(long_word,longest))
```

```python
#6
L=[]
with open("/content/drive/MyDrive/Ex_text_files/Ex_text_files/population.txt") as f:
  lines = [line.rstrip().split("\t") for line in f]
  for line in lines:
    if line[0].startswith("G") and int(line[1]) >=500000:
      L.append(line)
  for country in L:
    print(country[0])
```
