```python
# Input function
# The input function's job is to ask the user to type
# something in and to capture what the user types.
# The part in quotes is the prompt that the user sees.
# We use the float function to get numerical inputs.
```

```python
# E.g:
temp = float(input("Enter temperature: "))
# We need to give a name to the value that the user enters so that
# the program can remember it and use it in the second line.
# The name we use is temp and we use the equals sign to assign
# the user’s value to temp.
print(temp)
# The print function prints out the number.
```

```python
# Temperature converter program
temp = float(input("Enter temperature in Celsius: "))
print("The temperature in Fahrenheit is", (9/5)*temp + 32)
```

```python
# Program that computes the average of two numbers that the user enters:
num1= float(input("Enter number 1: "))
num2= float(input("Enter number 2: "))
print("The average of the two numbers is: ",(num1+num2)/2)
```

```python
# We use int for integers
num = int(input("Enter an integer: "))
print(num)
```

```python
# Getting input
# The input function is a simple way for your program to get
# information from people using your program.
# The basic structure is
# variable name = input(message to user)
# E.g
name = input('Enter your name: ')
print('Hello,', name)
# The above works for getting text from the user.
```

```python
# To get numbers from the user to use in calculations
# we do the following
num = float(input("Enter a number: "))
print('Your number squared:', num*num)
```

```python
# Printing
# The print function requires parenthesis around its arguments.
# E.g
print("Hello")
# In the example above, its only argument is the string 'Hello'
# Anything inside quotes will (with a few exceptions) be printed
# exactly as it appears.
# E.g
print("5 + 9")
print(5 + 9)
```

```python
# To print several things at once, separate them by commas.
# Python will automatically insert spaces between them.
# Below is an example and the output it produces.
print('The value of 5 + 9 is', 5 + 9)
print('B', 7, 'PQR', 23)
print("The value of 5 + 9 is",5 + 9,".")
```

```python
# Optional arguments: There are two optional arguments
# to the print function(sep which is short for separator and end).
# They are useful for making your output look nice.
```

```python
# usage of the optional argument sep
print("The value of 5 + 9 is",5 + 9,".", sep="")
```

```python
# Create a space between is and 14
print("The value of 5 + 9 is ",5 + 9,".", sep="")
```

```python
# Another way of creating a space between is and 14
print("The value of 5 + 9 is"," ",5 + 9,".", sep="")
```

```python
# end
# The print function will automatically advance to the next line.
# For instance, the following will print on two lines:
print("On the first line")
print("On the second line")
```

```python
# Usage of the second optional argument end
# It keeps the print function from advancing to the next line.
print("On the first line", end="")
print("On the second line")
```

```python
# Creating a space between line and on
print("On the first line", end=" ")
print("On the second line")
```
