```python
#For loop: It is used to repeat things over and over a finite
#number of times.
```

```python
#The structure of a for loop is as follows:
#for variable name in range(number of times to repeat):
#     statement to be repeated
```

```python
#Note:
#The word for must be in lowercase
#the first line must end with a colon
#the statements to be repeated must be indented
#Indentation is used to tell Python which statements will be repeated.
```

```python
#E.g1
#The following program will print Hello 5 times:
for k in range(5):
  print("Hello")
```

```python
#E.g2: Write a program that asks the user for an integer
# and print its square. Repeat this action 4 times
# then print that the loop is done.
```

```python
for k in range(4):
  num = int(input("Enter an integer: "))
  print("The square of your integer is",num*num)
print("The loop is now done.")
```

```python
#In the above program, the fourth line is not indented,
#so it is not part of the loop and only gets executed once,
#after the loop has completed.
```

```python
#E.g3: Write a program that prints A, then B,
#then it alternates C’s and D’s three times
#and then finishes with the letter E once.
```

```python
print("A")
print("B")
for j in range(3):
  print("C")
  print("D")
print("E")
```

```python
# The range function
# The value we put in the range function determines
# how many times we will loop
# range(p) produces a list of numbers from zero to p-1
#E.g
# range(4) produces 0,1,2 and 3.
# range(1,7) produces 1,2,3,4,5, and 6.
# range(4,8) produces 4,5,6, and 7.
# range(start,end) produces start,start + 1 up to end - 1.
# range(3,9,2) produces 3,5, and 7.
# range(11,3,-2) produces 11,9,7, and 5.
# range(start,end,step) produces start,start + step,
```

```python
# Here is an example program that counts down from 7.
for k in range(7,0,-1):
  print(k,end=" ")
```
