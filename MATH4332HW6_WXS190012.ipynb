```python
#Winston Shih (WXS190012)
#MATH 4332 HW 6
#1
def closest(numberlist, n):
  closest=numberlist[0]
  min=abs(numberlist[0]-n)
  for i in range(len(numberlist)):
    if abs(numberlist[i]-n)<min:
      min=abs(numberlist[i]-n)
      closest=numberlist[i]
    elif abs(numberlist[i]-n)==min:
      if numberlist[i]<closest:
        closest=numberlist[i]
  return closest
numlist=[1,9,11,5,6,7]
number=10
print(closest(numlist, number))
```

```python
#2
def names(s):
  words=s.split()
  l=[]
  for word in words:
    if word[0].isupper():
      if not word[-1].isalpha():
        word=word[:-1]
      l.append(word)
  return l
s = """Anna is 7 years old, and her sister Olivia is 2 years old.
Evelyn and Paul, their parents, have 3 kids."""
names_list=names(s)
print(names_list)
print(len(names_list))
```

```python
#3
def grades(f):
  scores=[line.strip() for line in f]
  for line in scores:
    record=line.split()
    firstname=record[0]
    lastname=record[1]
    lastname=lastname[:-1]
    score=record[2]
    if score=='B':
      print(firstname+' '+lastname)
file=open('/content/drive/MyDrive/FilePractice/HW6files/grades.txt')
grades(file)
file.close()
```

```python
#4
def logs():
  with open('/content/drive/MyDrive/FilePractice/HW6files/logdata.txt') as f:
    data=[line.strip() for line in f]
    dict=[]
    for line in data:
      record=line.split()
      host=record[0]
      user_name=record[2]
      time=record[3]
      time=time[:-1]
      time=time[1:]
      request=record[6]
      request=request[:-1]
      request=request[1:]
      dict.append({"host":host, "user_name":user_name, "time":time,
                   "request":request})
    return dict
log_list=logs()
for item in log_list:
  formatted_dict='{'+',\n'.join([f"'{key}': '{value}'" for key, value in item.items()])+'}'
  print(formatted_dict)
print(len(log_list))
```

```python
#5
def findall(s, l):
  list=[]
  s=s.lower()
  for i in range(len(s)):
    if s[i]==l:
      list.append(i)
  return list
print(findall("Avengers Assemble!", 'a'))
```

```python
#6
def digitalroot(n):
  sum=0
  while(n>0):
    sum+=n%10
    n=n//10
  return sum
print("The digital root of 198273 is",digitalroot(198273),sep=" ")
```

```python
#7
def closest(numberlist, n):
  largest=0
  for i in range(len(numberlist)):
    if numberlist[i]<n and numberlist[i]>largest:
      largest=numberlist[i]
  return largest
numlist=[60, 65, 67, 68, 69, 90, 100]
number=80
print(closest(numlist, number))
```
