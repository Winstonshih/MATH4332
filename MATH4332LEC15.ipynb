```python
# Some other useful methods of dictionary
d = {"A":100,"B":122}
# clear(): creates an empty dictionary
d.clear()
print(d)
```

```python
# update(): adds one or more items into existing
# dictionary
d = {"E":12,"F":122}
d.update({"H":{1:67},"S":[1,2]})
print(d)
```

```python
# pop(): removes item from dictionary
# and returns the value of its key
d = {'E': 12, 'F': 122, 'S': [1, 2], 'H': {1: 67}}
val = d.pop("S")
print(val)
print(d)
```

```python
# popitem(): removes recently inserted key-value
# pair and returns the pair
# d = {'E': 12, 'F': 122, 'H': {1: 67}}
v = d.popitem()
print(v)
print(d)
```

```python
# get(): Takes an argument which is a key
# It displays the value of given key if key is in
# dictionary and displays nothing if key not in dictionary
d = {'E': 12, 'F': 122, 'S': [1, 2]}
print(d.get("F"))
print(d.get("FE"))
```

```python
# setdefault(): Takes two arguments: key,value
# If key exists in dictionary, it returns its value
# otherwise it inserts the pair key:value into dictionary
# and returns key's value
d = {'E': 12, 'F': 122, 'S': [1, 2]}
print(d.setdefault("E",135))
print(d.setdefault("ED",35))
print(d)
```
