```python
# If statements
# Quite often in programs we only want to do something
# provided something else is true.
# Python’s if statement is what we need.
```

```python
# Guess-a-number program.
from random import randint
n = randint(1,5)
print(n)
guess = int(input("Enter a number between 1 and 5: "))
if guess == n:
  print("You won!")
```

```python
# In the above simple program, if the player guesses wrong,
# nothing happens. To fix that, We can add to the if statement
# an else statement as follows:
```

```python
from random import randint
n = randint(1,5)
print(n)
guess= int(input("Enter a number between 1 and 5: "))
if guess == n:
  print("You won!")
else:
  print("The correct number is",n)
```

```python
# Conditional operators
# The comparison operators are:
# ==(equal to. Make sure not to use = sign which is an assignment)
# >,<,>=,<=,!=(Not equal to)
```

```python
# E.g
x = float(input("Enter a number: "))
if x==7:
  print("x is equal to 7")
if x>=7:
  print("x is greater than or equal to 7")
if x>7:
  print("x is greater than 7")
if x<7:
  print("x is less than 7")
if x<=7:
  print("x is less than or equal to 7")
```

```python
# elif statement
# E.g
# Letter grade assignment program
grade = float(input("Enter your score: "))
if grade >= 90:
  print('A')
if grade >= 80 and grade < 90:
  print('B')
if grade >= 70 and grade < 80:
  print('C')
if grade >= 60 and grade < 70:
  print('D')
if grade < 60:
  print('F')
```

```python
# A more elegant way to do the above is:
```

```python
grade = float(input('Enter your score: '))
if grade >= 90:
  print('A')
elif grade >= 80:
  print('B')
elif grade >= 70:
  print('C')
elif grade >= 60:
  print('D')
else:
  print('F')
```

```python
#2
temp = float(input("Enter a temperature in Celsius: "))
if temp < -273.15:
  print("The temperature is invalid b/c it is below absoute zero.")
elif temp == -273.15:
  print("Temp is absolute zero")
elif temp < 0:
  print("The temp is below freezing.")
elif temp == 0:
  print("The temp is at the freezing point.")
elif temp < 100:
  print("The temp is in the normal range.")
elif temp == 100:
  print("The temp is at boiling point.")
else:
  print("The temp is above the boiling point.")
```
