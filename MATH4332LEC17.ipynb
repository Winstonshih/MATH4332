```python
# Find the largest marging of victory
games = [line.split() for line in lines]
L = []
for g in games:
  diff = int(g[2])-int(g[4])
  L.append(abs(diff))
print(max(L))
```

```python
# What game provided the largest margin of victory
games = [line.split() for line in lines]
largest_margin = None
for g in games:
  diff = abs(int(g[2])-int(g[4]))
  if largest_margin is None or diff > largest_margin:
    largest_margin = diff
    game_information = g
print(largest_margin)
print(game_information)
```

```python
# Or
# What game provided the largest margin of victory
games = [line.split() for line in lines]
L = []
for g in games:
  diff = int(g[2])-int(g[4])
  L.append((abs(diff),g))
L.sort(reverse=True)
print(L[:1])
print("largest margin is:{:3d}".format(L[0][0]))
print("Corresponding game info is: {}".format(L[0][1]))
```

```python
# Wordplay
# Consider the file wordlist.txt
f = open("/content/drive/MyDrive/Trial/pyth_text_files/wordlist.txt")
wordlist = [line.rstrip() for line in f ]
f.close()
print(wordlist[:5])
```

```python
# Print all three letter words
L =[]
for word in wordlist:
  if len(word)==3:
    L.append(word)
print(len(L))
print(L[:5])
print(L[-5:])
```

```python
# Print all the words that start with gn or kn
L = []
for word in wordlist:
  if word[:2] == "gn" or word[:2]=="kn":
    L.append(word)
print(len(L))
print(L[:5])
print(L[-5:])
```

```python
#OR
# Print all the words that start with gn or kn
L1 = []
for word in wordlist:
  if word[:2] in ["gn","kn"]:
    L1.append(word)
print(len(L1))
print(L1[:5])
print(L1[-5:])
```

```python
# Determine what percentage of words that starts with a vowel
vows = ["a","e","i","o","u"]
count = 0
for word in wordlist:
  if word[0] in vows:
    count+=1
print("Percentage is: {:.2f}".format(100*count/len(wordlist)))
```

```python
# Or
# Determine what percentage of words that starts with a vowel
count=sum([1 for word in wordlist if word[0] in ["a","e","i","o","u"]])
print("Percentage is: {:.2f}".format(100*count/len(wordlist)))
```

```python
# Print all 7-letter words that start with th and end with ly
L3 = []
for word in wordlist:
  if len(word)==7 and word[:2] == "th" and word[-2:]=="ly":
    L3.append(word)
print(L3)
```

```python
# Print the first ten words that start with q using a for loop
L = []
for word in wordlist:
  if word[0] == "q":
    L.append(word)
print(L[:10])
```

```python
# Or Print the first ten words that start with q using for/break/else
count = 0
l = []
for word in wordlist:
  if word[0] == "q":
    l.append(word)
    count +=1
  if count == 10:
    print("The list of the first ten words that start with q is:")
    break
else:
  print("There are less than 10 words that start with q. The list is:")
print(l)
```

```python
# Or Print the first ten words that start with q using while loop
# Words are listed in alphabetical order
i=0
while wordlist[i][0]!='q':
  i=i+1
print(wordlist[i:i+10])
```
