```python
# Dictionary
# A dictionary is a more general version of a list.
# To declare a dictionary we enclose it in curly braces, {}.
# Each entry consists of a pair separated by a colon.
# The first part of the pair is called the key and the second
# is the value. The key acts like an index.
```

```python
# Basics
# Creating a dictionary: set of pairs of keys(must be unique) values
d = {"A":2418,"B":4332}
print(d["A"])
print(d["B"])
```

```python
# Keys must be immutable
# values can be of any type
# You cannot create a dictionary with a key as a list
# or as a dictionary
# d = {[1,2,3]:23, "A":23}
# d = {{"A":2}:123,1:56}
```

```python
# You can mix different types of keys in the same dictionary
# and different types of values, too.
d = {1:{"A":2},'[1,2,3]':[1,2],23.12:"Math"}
print(d[1])
print(d['[1,2,3]'])
print(d[23.12])
```

```python
# Changing dictionaries
d = {"A":2418,"B":4332}
# To change d['A'] to 2419, do
d["A"]=2419
print(d)
```

```python
# To add a new entry "C":2415 to dictionary, do d["C"]=2415
d["C"]=2415
print(d)

# Note that this does not work with lists. Doing L[2]=2415 on
# a list with two elements would produce an index out of range error.
# But it does work with dictionaries.
```

```python
# To delete an entry from a dictionary, use del operator
del d["B"]
print(d)
```

```python
# An empty dictionary
# The empty dictionary is {}, which is the dictionary
# equivalent of [] for lists or '' for strings.
```

```python
# Important note:
# The order of items in a dictionary will not necessarily be
# the order in which you put them into the dictionary.
# Internally, Python rearranges things in a dictionary in
# order to optimize performance.
```

```python
# Dictionary Examples
# Dictionary of definition
d = {"A":"Def of A", "B":"Def of B", "C":"Def of C"}
word = input("Enter a word: ")
print("The def of",word,"is",d[word])
```

```python
# This is useful in a program that works with Roman numerals
numerals = {"I":1,"V":5,"X":10}
print(numerals["V"])
```

```python
# A dictionary provides a nice way to represent a deck of cards:
# Using list comprehension
deck = [{"value":i,"suit":c } for c in
 ['spades', 'clubs', 'hearts', 'diamonds'] for i in range(2,15)]
print(deck)
```

```python
# in operator: It is used to tell if something is a key in a
# dictionary
d = {"A":1325, "B":2332}
# Referring to a key that is not in the dictionary
# will produce an error
print(d["C"])
```

```python
# To prevent this error, we can use the in operator to check first
# if a key is in the dictionary before trying to use the key
if "C" in d:
  print(d["C"])
else:
  print("Key is not in dictionary")
```

```python
# Examples
d = {"A":"Def of A", "B":"Def of B", "C":"Def of C"}
word = input("Enter a word: ")
if word in d:
  print("The def of",word,"is",d[word])
else:
  print("word is not in dictionary")
```

```python
# You can also use not in to see if a key is not in the dictionary.
d = {"A":1325, "B":2332}
letter = input('Enter a letter: ')
if letter not in d:
  print(letter,'not in dictionary')
else:
  print('The value of',letter,"is", d[letter])
```

```python
# Use d.items() to find all keys in dictionary
# into a list
d = {"A":12,"B":31,"C":77}
L=[]
for k in d.items():
  L.append(k[0])
print(L)
```

```python
# Use list comprehension to solve the above question
L = [k[0] for k in d.items()]
print(L)
```

```python
# Use d.items() to find all keys in dictionary
# that correspond to value 31
d = {"A":12,"B":31,"C":77,'[1,2,3]':31}
L=[]
for k in d.items():
  if k[1]==31:
    L.append(k[0])
print(L)
```

```python
# Use list comprehension to solve the above question
L = [k[0] for k in d.items() if k[1]==31]
print(L)
```

```python
# Use d.items() to find all values in dictionary
d = {"A":12,"B":23,"C":7}
L = []
for k in d.items():
  L.append(k[1])
print(L)
```

```python
# Use list comprehension to solve the above
L = [k[1] for k in d.items()]
print(L)
```

```python
# Another way to create a dictionary
d = dict([("A",123),("B",[2,23])])
print(d)

# This way of building a dictionary is useful if
# your program needs to construct a dictionary while it is running.
```

```python
# Dictionary comprehension
# It works similarly to list comprehensions.
# key=name, value=len(name)
names = ["Sam","Paul","Robert"]
d = {s:len(s) for s in names}
print(d)
```

```python
# Counting words
# We can use dictionaries to count how frequently
# certain words appear in a text.
# Most common names
names = ["Paul","Anna","Paul","Anna","Sam"]
d = {}
for name in names:
  if name in d:
    d[name]+=1
  else:
    d[name]=1
print(d)
```

```python
# Use the get method for dictionary to solve the
# above problem
names = ["Paul","Anna","Paul","Anna","Sam"]
d = {}
for name in names:
  d[name]=d.get(name,0) + 1
print(d)
```

```python
# Use count method to solve the above
# above problem
names = ["Paul","Anna","Paul","Anna","Sam"]
d = {}
for name in names:
  if name not in d:
    d[name]=names.count(name)
print(d)
```

```python
# Frequency of each word of the following string
s=""" the clown ran after the car and the car ran into the
tent and the tent fell down on the clown and the car
"""
```

```python
# Here, the string does not contain any punctuation
# Create a list of words of s
L = s.split()
print(L)
```

```python
# Create a dictionary of frequencies of the words of s
d = {}
for item in L:
  d[item]= d.get(item,0) + 1
print(d)
```

```python
# Find the 3 most common words of s
# Create a list of (value,key) pairs
L1 = []
for key, val in d.items():
  pair = (val, key)
  L1.append(pair)
print(L1)
```

```python
# Sort the list L1 in descending order
L1.sort(reverse=True)
print(L1)
```

```python
# Observe that by creating the list of (key,value) pairs
L2 = []
for key, val in d.items():
  pair = (key, val)
  L2.append(pair)
print(L2)
```

```python
# The sort methods only sorts elemnts of the list
# with respect to the first component of the tuples
# which are strings here, therefore words will be sorted
# same way words are sorted in the english dictionary
```

```python
# Sort the list L2 in descending order will produce a different
# result than the one we are looking for
L2.sort(reverse=True)
print(L2)
```

```python
# Back to L1
# Most 3 common words and their frequencies
for item in L1[:3]:
  print(item[1],item[0])
```

```python
#Or
for val,key in L1[:3]:
  print(key,val)
```
