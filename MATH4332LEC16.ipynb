```python
# Text Files

# There is a ton of interesting data to be found on the internet
# stored in text files. In this chapter we will learn how to work
# with data stored in text files.
```

```python
# Reading from files:
# Suppose we have a txt file called file.txt
# and we want to read its content into python.
# The content of the file is:
"""
Hello.
This is a text file.
Bye!
"""
```

```python
# There are svereral ways to do this. Let us look at some of them.
# First way uses list comprehension to load the file line by line
# into a list
lines = [line.strip() for line in open("file.txt")]
```

```python
# The string method strip removes any whitespace characters = ( ,\n,\t)
# from the beginning and the end of the line
# The list named lines now is:
lines = ["Hello.","This is a text file.","Bye!" ]
```

```python
# Note: use rstrip if you need to preserve whitespace characters at
# the beginning of the line
# Let us remove strip
lines = [line for line in open("file.txt")]
# The list lines would be
lines = ["Hello.\n","This is a text file.\n","Bye!" ]
```

```python
# Second way of reading a text file loads the entire file
# into a string
s = open("file.txt").read()
# The string s now is
s= "Hello.\nThis is a text file.\nBye!"
```

```python
# Third way
with open("file.txt") as f:
  data = f.read()
data= "Hello.\nThis is a text file.\nBye!"
```

```python
# Examples of reading a file
f = open("file.txt")
# f is a file handle or file object that represents the file.
# The content of the file is stored in f
for line in f:
  print(line)
f.close()
```

```python
# We can remove the empty lines between the two lines
f = open("file.txt")

for line in f:
  line = line.rstrip()
  print(line)
f.close()
```

```python
# read(): reads all content of the file
f = open("file.txt").read()
print(f)
```

```python
# readline(): It reads the content of a single line
f = open("file.txt")
print(f.readline().rstrip())
print(f.readline().rstrip())
f.close()
```

```python
# Using while loop and readline() to read all content of a file
# line by line
f = open("file.txt")
while True:
  line = f.readline()
  if line == "":
    break
  line=line.rstrip()
  print(line)
f.close()
```

```python
# Examples

# 1 temps.txt file
# write a program that reads a list of temperatures from a file
# called temps.txt, converts those temperatures to F
# and write the results to a file called ftemps.txt
```

```python
# Answer
f = open("ftemps.txt","w")
lines = [line.rstrip() for line in open("/content/drive/MyDrive/FilePractice/pyth_text_files/temps.txt")]
for line in lines:
  conv = float(line)*9/5 + 32
  print("{:.2f}".format(conv), file=f)
f.close()
```

```python
# Accessing elements of ftemps.txt
lines = [line.rstrip() for line in open("/content/ftemps.txt")]
print(lines)
```

```python
# Use writelines to solve the above problem:
f = open("nftemps.txt","w")
lines = [line.rstrip() for line in open("/content/drive/MyDrive/FilePractice/pyth_text_files/temps.txt")]
L = []
for line in lines:
  conv = float(line)*9/5 + 32
  L.append("{:.2f}\n".format(conv))
f.writelines(L)
f.close()
```

```python
# Another method to solve the above problem:
f = open("/content/drive/MyDrive/FilePractice/pyth_text_files/temps.txt", "r")
fh = open("Nftps.txt","w")
for line in f:
  conv = float(line)*9/5 + 32
  print("{:.2f}".format(conv), file=fh)
fh.close()
f.close()
```
