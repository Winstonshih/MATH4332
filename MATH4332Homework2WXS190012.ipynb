```python
#3a
import math
from math import *
print("Enter 10 test scores.")
smallest=1000
secondsmallest=1000
overhundred=0
sum=0
count=0
for i in range(10):
  score=int(input("Enter a score: "))
  if score<smallest:
    secondsmallest=smallest
    smallest=score
  elif score<secondsmallest:
    secondsmallest=score
  if score>100:
    overhundred+=1
  sum+=score
  count+=1
print("The smallest number:",smallest)
print("The second smallest number:",secondsmallest)
#3b
if(overhundred>0):
  print("A score over 100 has been entered!")
#3c
sum-=(smallest+secondsmallest)
count-=2
print("Average after dropping 2 lowest scores:",sum/count)
```

```python
#4
import random
from random import randint
print("Guess 5 numbers between 1 and 10.")
print("You get 10 points for each right guess.")
print("You lose a point for each wrong guess.")
score=0
for i in range(5):
  num=randint(1,10)
  guess=int(input("Make a guess: "))
  if(guess==num):
    score+=10
  elif(guess!=num):
    score-=1
  print("The number is",num)
print("Score:",score,sep=" ")
```

```python
#5b
decrypt=input("Enter a string to decrypt: ")
if len(decrypt)%2==0:
  mid=len(decrypt)//2
elif len(decrypt)%2==1:
  mid=(len(decrypt)+1)//2
decrypt1=decrypt[:mid]
decrypt2=decrypt[mid:]
decryptedmessage=""
for i in range(mid):
  if i<len(decrypt2):
    decryptedmessage+=decrypt1[i]
    decryptedmessage+=decrypt2[i]
  else:
    decryptedmessage+=decrypt1[i]
print("The decrypted message is:",decryptedmessage)
```
