```python
# Basics:
# Functions are defined with the def statement.
# The statement ends with a colon, and the code that is part of
# the function is indented below the def statement.
# def name_function():
#     indented codes
```

```python
# Suppose you need to print a box of stars
# like the one below at several points in your program.
# *********
# *       *
# *       *
# *********
```

```python
# Put the code into a function, and then whenever you need a box
# just call the function rather than typing several lines of
# redundant code. Here is the function.
def box():
  print("*"*9)
  print("*"," "*7,"*",sep="")
  print("*"," "*7,"*",sep="")
  print("*"*9)
```

```python
# Calling the function box()
box()
```

```python
# We can change the size of the box
# by modifying the code inside the function
def box():
  print("*"*17)
  print("*"," "*15,"*",sep="")
  print("*"," "*15,"*",sep="")
  print("*"*17)
```

```python
box()
```

```python
# Returning values
# We can write functions that perform calculations and return a result
```

```python
# Example
# Simple function that converts temperature from C to F
def convert1(t):
  return t*9/5 + 32
```

```python
# Calling the function convert1()
print(convert1(21.33))
```

```python
def convert2(t):
  f = t*9/5 + 32
  return f
```

```python
print(convert2(21.33))
```

```python
def convert3(t):
  f = t*9/5 + 32
  result = "{:.2f}".format(f)
  return float(result)
```

```python
print(convert3(21.33))
```

```python
# The return statement is used to send the result
# of a function’s calculations back to the caller.
# Notice that the function itself does not do any printing.
# The printing is done outside of the function.
# That way, we can do math with the result.
```

```python
# Return allows doing some math operations
print(convert1(34))
print(convert1(34)*2)
print(convert1(34)+25)
```

```python
# If we had just printed the result in the function instead
# of returning it, the result would have been
# printed to the screen and forgotten about,
# and we would never be able to do anything with it.
```

```python
def convert11(t):
  print( t*9/5 + 32 )
```

```python
print(convert11(34)+25)
```

```python
print(convert11(34)*2)
```

```python
# function can return multiple values as a list
# We want to write a function that solves the 2 by 2 system below
# ax + by =e and cx + dy=f
# assume that ad-bc!=0 ie the system has a unique solution
# By Cramer's rule we have:
# x = (e*d-b*f)/(a*d-b*c), y=(a*f-e*c)/(a*d-b*c)
# We want our function to return both x and y
```

```python
def solve(a,b,c,d,e,f):
  x = (e*d-b*f)/(a*d-b*c)
  y =(a*f-e*c)/(a*d-b*c)
  return [x,y]
```

```python
# xsol,ysol = solve(a,b,c,d,e,f)
# print("The solution is x=",xsol,"and y=",ysol)
# print("The solution is x= {} and y = {}".format(xsol,ysol))
```

```python
# Example:
# 2x - y = 1, x + 2y = 2
solve(2,-1,1,2,1,2)
```

```python
xsol,ysol = solve(2,-1,1,2,1,2)
print("The solution is x = {} and y = {}".format(xsol,ysol))
```

```python
# Keywords arguments
def s_d(x,y,z):
  return x-y-z
# Every time you call this function, you have to remember
# the correct order of the arguments.
```

```python
# arguments must be given in correct order x, y and z
s_d(7,6,9)
```

```python
s_d(6,9,7)
```

```python
# Fortunately, Python allows us to name the arguments when calling
# the function, as shown below:
s_d(x=7,y=6,z=9)
```

```python
s_d(z=9,x=7,y=6)
```

```python
# As we can see, the order of the arguments does not matter
# when you use keyword arguments.
```

```python
# Global variable: Variable available to multiple functions
```

```python
a = 7 # Golbal variable
def smt():
  print("Inside function:",a)
smt()
print("Outside function:",a)
```

```python
# If global and local variables have the same name
# first preference will be given to local variable
```

```python
a = 7 # Golbal variable
def smt():
  a = 15 # local variable
  print("Inside function:",a)
smt()
print("Outside function:",a)
```

```python
# Create global variable inside the function
a = 7 # Global variable
def smt():
  global a
  a = 13
  print("Inside function:",a)
smt()
print("Outside function:",a)
```

```python
a = 7 # Global variable
def smt():
  a = 13 # Local variable
  x = globals()["a"] # Access global variable inside
  print("Inside function:",a,x)
smt()
print("Outside function:",a)
```

```python
# Local variables cannot be accessed outside the function
```

```python
def lcv():
  b = 15
  print(b)
lcv()
print(b)
```

```python
a = 7 # Global variable
def smt():
  a = 13 # Local variable
  x = globals()["a"] # Access global variable inside
  print("Inside function:",a,x)
smt()
# x is a local variable in the function smt()
print("Outside function:",a,x)
```
