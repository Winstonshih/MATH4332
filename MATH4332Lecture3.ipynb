```python
# Python Modulo Operator Basics
# The modulo operator, like the other arithmetic operators,
# can be used with the numeric types int and float.
```

```python
# Modulo Operator With int
# a%b returns the remainder of standard Euclidean division
```

```python
print(11%3)
print(17%12)
```

```python
# Modulo Operator With a Negative Operand
# When a negative operand is introduced, r is determined by:
# r = a - (b*floor(a/b))
```

```python
print((-11)%3)
# r = -11 - (3*floor(-11/3))
# r = -11 - (3*floor(-3.6666))
# r = -11 - (3*-4)
# r = -11 - (-12) = -11 + 12 = 1
```

```python
print(11%(-3))
# r = 11 - (-3*floor(11/-3))
# r = 11 - (-3*floor(-3.6666))
# r = 11 - (-3*-4) = 11 - 12 = -1
```
