```python
# Lists
# Creating a list
# list of 3 items (items are separated by commas)
L = [2,3,11]
print(L)
# Empty list
L = []
print(L)
# Empty list
H = list()
print(H)
# Long list: you can split it across several lines
F = [1,3,5
     ,3,45,
     67,49]
print(F)
```

```python
# Data types:
# Lists can contain all kinds of things,
# even other lists.
# E.g
L = [1,"ab",3.14, [1,2,7]]
# Length of L: len(L)
print(len(L))
```

```python
# Similarities to strings:
# len, in, indexing and slicing
# methods: index and count
# + and *
# Looping
```

```python
# E.g
L = [1,3,-7,23,27,45]
# Find the number of items in L
print(len(L))
# Check if a list contains an item
L = [1,3,-7,23,27,45]
if 3 in L :
  print("The lis contains 3")
if 7 not in L:
  print("The list does not contain 7")
```

```python
# Indexing and slicing
L = [12,34,56,78,99,[1,2,3],"abc"]
# First element of the list
print(L[0])
# Last element of the list
print(L[-1])
# First two elements : L[:2]
print(L[:2])
# Last 3 elements of the list: L[-3:]
print(L[-3:])
```

```python
# + operator: L1 + L2 adds L2 at the end of L1
L1 = [1,2]
L2 = [99,"abcd",[12,34,45]]
# L1 + L2 = [1,2,99,"abcd",[12,34,45]]
print(L1 + L2)
```

```python
# * operator: It repeats the list
# L1*2 = [1,2,1,2]
print(L1*2)
```

```python
# Looping through a list
L = [99,"abcd",[12,34,45]]
for item in L:
  print(item , end = " ")
print()
for k in range(len(L)):
  print(L[k], end=" ")
print()
```

```python
# Useful built in functions
# len : returns the number of elements in the list
# For list of numbers:
# sum : returns the sum of the item in the list
# min : returns the minimum of the list
# max : returns the maximum of the list
# average : returns sum(L)/len(L)
```

```python
# List methods
# append(x): adds x to the end of the list
# sort() : it sorts the list
# count(x): number of occurrences of x in the list
# index(x): location of first occurrence of x in the list
# reverse() : reverses the list
# remove(x): remove first occurence of x from the list
# pop(k): removes item at index k and returns its value
# insert(k,x): inserts x at index k of the list
# pop(): removes last item in the list and returns its value
# extend(): it takes list as argument and appends all of its items
```

```python
# Examples
L = [23, 89, 37, -99]
L.sort() # it sorts the list inplace in increasing order
print(L)
```

```python
L.sort(reverse=True) # it sorts in decreasing order
print(L)
```

```python
# Important note
# String methods do not change the original string,
# but list methods do change the original list.
```

```python
#E.g: We want to change the original string
s = "abcdef"
# wrong way
s.replace("b","x")
print(s)
```

```python
# right way
s = s.replace("b","x")
print(s)
```

```python
# We want to sort the list
L = [23, 89, 37, -99]
# wrong way
L=L.sort()
print(L)
```

```python
# right way
L = [23,89,37,-99]
L.sort()
print(L)
```

```python
# Other list methods
# Making copies of lists: H = L[:]
L = [1,2,"abc"]
H = L[:]
H.append("last")
print(L)
print(H)
```

```python
# same here
L = [1,2,"abc"]
K = L.copy()
K.append(34)
print(L)
print(K)
```

```python
# Not a good way to copy a list
L = [1,2,"abc"]
F = L
F.append("last")
print(L)
print(F)
```

```python
# Changing a list
L = [23,77,89,25]
# Replace first element of L with 12.23
L[0]= 12.23
print(L)
# Insert 100 at index 2
L.insert(2,100)
print(L)
```

```python
# del : Removes items by index or slice
L = [23,77,89,25]
del L[1]
print(L)
# Delete multiple elements with slice
L = [23,77,89,25]
del L[1:3]
print(L)
# clear(): Remove all items from the list
L = [23,77,89,25]
L.clear()
print(L)
```
