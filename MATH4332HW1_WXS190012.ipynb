```python
#3
temperature=float(input("Enter a temperature: "))
units=int(input("Enter emperature unit (Celsius[0] or Fahrenheit[1]): "))
if units==0:
  converted_temperature=(9*temperature/5)+32
  print(temperature,"°C is equal to ",converted_temperature," °F.",sep="")
if units==1:
  converted_temperature=(5*(temperature-32))/9
  print(temperature,"°F is equal to ",converted_temperature," °C.",sep="")
```

```python
#4
credits=int(input("Enter the number of credits you have taken: "))
if credits<=23:
  print("You are a freshman.")
if credits>=24 and credits<=53:
  print("You are a sophomore.")
if credits>=54 and credits<=83:
  print("You are a junior.")
if credits>=84:
  print("You are a senior.")
```
